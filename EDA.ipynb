```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Books_Data_Clean.csv")
```

```python
df
```

```python
df.head(5)
```

```python

df.describe()
```

```python
df = df[df["Publishing Year"] >1900]
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
# df.dropna(subset = "Book Name")
df.dropna(subset = "Book Name", inplace = True)
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df.nunique()
```

```python
plt.hist(df["Publishing Year"])

plt.xlabel("Publishing Year")
plt.ylabel("Frequency")
plt.title("Publishing Frequency Distribution")

plt.show()
```

```python
df["genre"].value_counts().plot(kind = "bar")

plt.xlabel("Genre")
plt.xlabel("Books")
plt.title("Books in each Genre")

plt.show()
```

```python
df.groupby("Author")["Book_average_rating"].mean()
```

```python
# type(df.groupby("Author")["Book_average_rating"].mean())
df.groupby("Author")["Book_average_rating"].mean().sort_values(ascending = False)
```

```python
sns.boxplot(data=df,x="genre",y="Book_ratings_count", hue="genre")
plt.xlabel("Genre")
plt.ylabel("Books Rating Count")
plt.title("Ratings in each Genre")

plt.show()
```

```python
sns.scatterplot(df, x="sale price", y="units sold", hue="genre")

plt.xlabel("Sale Price")
plt.ylabel("Units Sold")
plt.title("Sales and Units Scatter Plot")

plt.show()
```

```python
# language_counts = df["language_code"].value_counts().sort_values()
language_counts = df["language_code"].value_counts()
language_counts
```

```python
plt.pie(language_counts, labels=language_counts.index, startangle=90, autopct = "%1.1f%%")

plt.title("Language Pie Chart")

plt.show()
```

```python
plt.scatter(df["Book_average_rating"],df["Book_ratings_count"])

plt.xlabel("Book Average Rating")
plt.ylabel("Book Rating Count")
plt.title("Average Book Rating")

plt.show()
```

```python
gross_sales=df.groupby("Author")["gross sales"].sum()
```

```python
sns.boxplot(df,x="Author_Rating",y="units sold", hue="genre")

plt.xlabel("Author Ratings")
plt.ylabel("Units Sold")
plt.title("Author Ratings v Units Sold")
```

```python
df.groupby("Publishing Year")["units sold"].sum().plot(kind="line", marker = ".")
plt.xlabel("Publishing Year")
plt.ylabel("Units Sold")
plt.title("Units Sold in Each Year")

plt.show()
```
